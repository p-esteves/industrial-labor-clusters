# Camada 1: Econometria Espacial & Geografia Econômica (PySAL & GeoPandas)

Este notebook executa o teste de autocorrelação espacial global ($I$ de Moran) e a identificação de clusters locais de produtividade industrial (LISA Hotspots/Coldspots) para as 27 Unidades Federativas do Brasil.

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
from src.spatial.spatial_weights import load_uf_geodataframe, create_spatial_weights
from src.spatial.moran_analysis import calculate_global_moran
from src.spatial.lisa_clustering import calculate_lisa_clusters
from src.viz.spatial_plots import plot_moran_scatterplot, plot_lisa_map
```

## 1. Carregamento dos KPIs por UF e Geometria Vetorial

```python
df_kpis = pd.read_csv('../data/outputs/kpis_summary.csv', sep=';')
df_kpis.columns = [c.lower() for c in df_kpis.columns]
gdf = load_uf_geodataframe(df_kpis)
gdf.head()
```

## 2. Construção da Matriz de Pesos Espaciais ($W$)

```python
w, gdf_ordered = create_spatial_weights(gdf, method='knn', k=4, row_standardize=True)
print(f'Matriz W construída para {w.n} observações com {w.k} vizinhos mais próximos.')
```

## 3. Autocorrelação Espacial Global ($I$ de Moran)

```python
target_var = 'produtividade_r_por_trabalhador'
moran_results = calculate_global_moran(gdf_ordered, target_var, w, permutations=999)
print('I de Moran Global:', moran_results['moran_i'])
print('P-valor:', moran_results['p_value'])
print('Padrão:', moran_results['pattern'])
```

## 4. Indicadores Locais de Associação Espacial (LISA)

```python
df_lisa, lisa_counts = calculate_lisa_clusters(gdf_ordered, target_var, w, p_threshold=0.10)
print('Contagem dos Quadrantes LISA:', lisa_counts)
fig_map = plot_lisa_map(df_lisa, lisa_label_col='lisa_label')
```
